# DATA CLEANING

```python
import pandas as pd
```

## Loading Data

```python
df=pd.read_csv("data/resumes.csv")
df
```

## Cleaning

```python
try:
    from nltk.corpus import stopwords
    STOPWORDS = set(stopwords.words("english"))
except Exception:
    from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
    STOPWORDS = set(ENGLISH_STOP_WORDS)
```

```python
import re
_URL_RE = re.compile(r"http\S+|www\.\S+")
_HTML_RE = re.compile(r"<.*?>")
_NONALPHA_RE = re.compile(r"[^a-z\s]")
_MULTISPACE_RE = re.compile(r"\s+")
def clean_text(text,stopwords=True,min_tokens=2):
    text=str(text).lower()
    text = str(text).lower()
    text = _URL_RE.sub(" ", text)
    text = _HTML_RE.sub(" ", text)
    text = _NONALPHA_RE.sub(" ", text)
    tokens = _MULTISPACE_RE.sub(" ", text).strip().split()
    res = [t for t in tokens if len(t) >= min_tokens and not (stopwords and t in STOPWORDS)]
    return " ".join(res)
```

```python
df['clean_text']=df['text'].astype(str).map(clean_text)
df=df[df['clean_text'].str.strip()!='']
print('after cleaning:',len(df),'rows')
```

## Comparision

```python
ex=df.iloc[0]
print('RAW  :',str(ex['text'])[:200].replace(chr(10),' '))
print('\nCLEAN:',ex['clean_text'][:200])
```

```python
for lab in df['label'].unique()[:6]:
    uc=df[df['label']==lab]['text'].iloc[0]
    c=df[df['label']==lab]['clean_text'].iloc[0]
    print(f'RAW  -  [{lab}] '+str(uc)[:160].replace(chr(10),' ')+'...\n')
    print(f'CLEAN  -  [{lab}] '+str(c)[:160].replace(chr(10),' ')+'...\n')
```

## save cleaned text

```python
df[['clean_text','label']].to_csv('data/resumes_cleaned.csv',index=False)
print('saved data/resumes_cleaned.csv',df.shape)
```
