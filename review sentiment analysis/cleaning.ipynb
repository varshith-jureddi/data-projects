# Text Cleaning

```python
import pandas as pd
import utils
```

```python
df=utils.load_data()
print('rows:',len(df))
```

```python
df['clean_text']=df['text'].astype(str).map(utils.clean_text)
df=df[df['clean_text'].str.strip()!='']
print('after cleaning:',len(df),'rows')
```

```python
ex=df.iloc[0]
print('RAW  :',str(ex['text'])[:200].replace(chr(10),' '))
print('\nCLEAN:',ex['clean_text'][:200])
```

```python
df[['clean_text','label']].to_csv('data/reviews_cleaned.csv',index=False)
print('saved data/reviews_cleaned.csv',df.shape)
```
